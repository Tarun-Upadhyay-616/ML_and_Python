```python
from sklearn.datasets import make_blobs
import seaborn as sns
```

```python
X,y = make_blobs(
    n_samples = 1000,
    n_features=2,
    centers = 3
)
```

```python
y
```

```python
sns.scatterplot(x = X[:,0] , y=X[:,1] )
```

```python
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans
```

# Elbow Method

```python
wcss = []
for k in range(1,21):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x = range(1,21),y = wcss)
```

### so we get k=2 from Elbow Method

```python
ss = []
for k in range(2,21):
    kmeans = KMeans(n_clusters=k)
    labels = kmeans.fit_predict(X)
    ss.append(silhouette_score(X,labels))
```

```python
import matplotlib.pyplot as plt
```

```python
ax = sns.lineplot(x=range(2,21),y=ss ,marker = "o")
for x, y in zip(range(2,21), ss):
    # ax.text(x_coordinate, y_coordinate, label_text)
    # va='bottom' keeps the text above the marker
    ax.text(x, y + 0.02, f'({x}, {y:.2f})', 
            horizontalalignment='center', 
            size='small', 
            color='black', 
            weight='semibold')

plt.show()
```
